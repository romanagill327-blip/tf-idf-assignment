## 1. Import libraries

```python
import pandas as pd
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity

print("Libraries imported successfully.")
```

## 2. Load the 10 documents and check word counts

```python
titles = ['AI in Healthcare', 'Climate Change', 'Basketball Strategy', 'Italian Pasta', 'Space Exploration', 'Cybersecurity', 'Music Theory', 'Gardening Vegetables', 'Economics Inflation', 'Public Health']

documents = ['Artificial intelligence is transforming healthcare by helping doctors analyze medical images, predict disease risk, and personalize treatments. Machine learning models can scan thousands of patient records to find patterns that humans might miss. Hospitals use AI to reduce waiting times and improve diagnosis accuracy. However, AI systems require large, clean datasets and careful validation before clinical use. Privacy, bias, and accountability are major concerns.\nIf used responsibly, AI can support clinicians rather than replace them, making healthcare faster, safer, and more equitable for diverse populations.', 'Climate change is driven by greenhouse gas emissions from burning fossil fuels. Renewable energy sources such as solar, wind, and hydropower can reduce carbon pollution. Governments and businesses are investing in batteries, smart grids, and energy efficiency. Communities also need adaptation plans for heat waves, floods, and droughts. Climate policy involves trade-offs between economic growth and environmental protection. Individual actions matter, but large-scale systemic change is essential.\nAccurate data and international cooperation help track progress toward lowering emissions and limiting global temperature rise.', 'Basketball strategy depends on spacing, passing, and player movement. Coaches design plays to create open shots and mismatches. Offenses use pick-and-rolls, screens, and fast breaks, while defenses switch, trap, or zone. Analytics now influences decisions about shot selection, lineup combinations, and pace. A team with strong chemistry often outperforms a group of individual stars. Rebounding and turnovers frequently decide close games.\nEffective communication on defense is as important as scoring ability. Successful teams practice repeatedly and adjust tactics based on opponent weaknesses and game situations.', "Italian pasta cooking begins with quality ingredients and simple techniques. A good sauce balances tomatoes, olive oil, garlic, herbs, and sometimes meat or cheese. Pasta should be cooked in salted water until al dente, then finished in the sauce for better flavor. Fresh basil and parmesan add aroma and richness. Different shapes suit different sauces: long pasta for light sauces, short tubes for chunky sauces. Regional traditions vary widely across Italy.\nHome cooks can improve by tasting, seasoning gradually, and respecting the pasta's texture rather than relying only on a timer.", 'Space exploration has advanced through robotic missions and human spaceflight. Mars rovers study geology, atmosphere, and signs of ancient water. Telescopes observe distant galaxies, exoplanets, and black holes. Space agencies and private companies are developing reusable rockets to lower launch costs. Astronauts face radiation, isolation, and microgravity health risks. Future missions may establish lunar bases or send crews to Mars. International collaboration and data sharing accelerate discovery.\nExploring space also produces technologies that benefit life on Earth, from satellite navigation to medical devices and materials science.', 'Cybersecurity protects computers, networks, and personal data from attacks. Common threats include phishing emails, malware, ransomware, and weak passwords. Multi-factor authentication and regular software updates reduce risk. Organizations use firewalls, encryption, and monitoring systems to detect intrusions. Employees are often the weakest link, so security training is essential. Data breaches can damage finances and reputation.\nStrong cybersecurity requires layered defenses, incident response plans, and privacy laws. As attackers become more sophisticated, defenders must continuously learn new tools, share threat intelligence, and design systems that are secure by default.', 'Music theory explains scales, chords, rhythm, and harmony. Musicians practice intervals, chord progressions, and ear training to perform and compose. Different genres use distinct rhythms and instrumentation. A melody becomes memorable through repetition, contrast, and tension. Improvisation requires both theoretical knowledge and listening skills. Recording software allows artists to edit, mix, and produce tracks at home.\nLive performance depends on timing, dynamics, and communication with bandmates. Learning music improves memory, coordination, and emotional expression. Consistent practice, feedback, and experimentation help musicians develop a personal style over time.', 'Gardening vegetables requires healthy soil, sunlight, water, and patience. Compost adds nutrients and improves soil structure. Gardeners rotate crops to prevent pests and diseases. Tomatoes, peppers, beans, and lettuce grow well in many home gardens. Mulch conserves moisture and reduces weeds. Raised beds help control drainage and soil quality. Pests can be managed with beneficial insects, companion planting, and careful monitoring. Harvesting at the right time improves flavor and nutrition.\nA successful garden depends on observing local climate, planning seasons, and learning from each growing cycle year after year.', 'Economics studies how people, firms, and governments allocate scarce resources. Inflation measures the rising cost of goods and services over time. Central banks adjust interest rates to influence borrowing, spending, and investment. High inflation reduces purchasing power, especially for low-income households. Unemployment and economic growth are key indicators of national health. Fiscal policy involves taxation and government spending.\nGlobal trade, supply chains, and energy prices affect local economies. Understanding incentives, markets, and data helps policymakers design stable policies. Economic decisions always involve trade-offs between efficiency, equity, and long-term sustainability.', 'Public health focuses on preventing disease and improving community well-being. Sleep, exercise, and nutrition strongly influence physical and mental health. Regular activity strengthens the heart, muscles, and immune system. A balanced diet includes vegetables, fruits, whole grains, and protein. Chronic stress can harm sleep, mood, and metabolism. Vaccination, clean water, and health education reduce illness. Public health campaigns encourage healthy habits and early screening.\nAccess to care and social support also shape outcomes. Small daily choices, supportive environments, and evidence-based policies can help populations live longer, healthier, and more productive lives.']

assert len(titles) == 10
assert len(documents) == 10

word_counts = [len(document.split()) for document in documents]
document_summary = pd.DataFrame({
    "Title": titles,
    "Word Count": word_counts
})
display(document_summary)

print(f"Total documents loaded: {len(documents)}")
print(f"Every document has more than 30 words: {all(count > 30 for count in word_counts)}")
```

## 3. Build the TF-IDF matrix

```python
# stop_words="english" removes common words such as "the", "is", and "and".
vectorizer = TfidfVectorizer(stop_words="english")
tfidf_matrix = vectorizer.fit_transform(documents)

feature_names = vectorizer.get_feature_names_out()
print("TF-IDF matrix shape (documents, unique terms):", tfidf_matrix.shape)
print("Vocabulary size:", len(feature_names))

assert tfidf_matrix.shape[0] == 10
assert tfidf_matrix.shape[1] == len(feature_names)
```

## 4. View the TF-IDF matrix as a DataFrame

```python
tfidf_df = pd.DataFrame(
    tfidf_matrix.toarray(),
    index=titles,
    columns=feature_names
)

print("Sample: first 3 documents and first 10 terms")
display(tfidf_df.iloc[:3, :10].round(3))
```

## 5. Find the top 5 TF-IDF terms for every document

```python
for title in tfidf_df.index:
    top_five = tfidf_df.loc[title].sort_values(ascending=False).head(5)
    print(f"\n{title}")
    display(top_five.rename("TF-IDF score").to_frame().round(4))
```

## 6. Query similarity challenge

```python
def rank_documents(query):
    # Convert the query to the same TF-IDF feature space as the documents.
    query_vector = vectorizer.transform([query])

    # Compare the query with all documents using cosine similarity.
    scores = cosine_similarity(query_vector, tfidf_matrix).flatten()

    # Return documents from highest similarity to lowest.
    ranked_indices = scores.argsort()[::-1]
    return pd.DataFrame({
        "Document": [titles[i] for i in ranked_indices],
        "Cosine Similarity": [scores[i] for i in ranked_indices]
    })

queries = [
    "renewable energy and climate policy",
    "password security and phishing",
    "healthy sleep exercise and diet"
]

for query in queries:
    print(f"\nQuery: {query}")
    display(rank_documents(query).round(4))

# Check that the expected document ranks first for each query.
expected_first = [
    "Climate Change",
    "Cybersecurity",
    "Public Health"
]
for query, expected in zip(queries, expected_first):
    actual = rank_documents(query).iloc[0]["Document"]
    print(f"Query '{query}' -> first result: {actual}")
    assert actual == expected, f"Expected {expected}, but got {actual}"

print("All three expected query rankings passed.")
```

## 7. Optional challenge: use unigrams and bigrams

```python
# ngram_range=(1, 2) includes single words and two-word phrases.
# max_df=0.8 removes terms found in more than 80% of documents.
# min_df=2 keeps terms that appear in at least 2 documents.
bigram_vectorizer = TfidfVectorizer(
    stop_words="english",
    ngram_range=(1, 2),
    max_df=0.8,
    min_df=2
)

bigram_matrix = bigram_vectorizer.fit_transform(documents)
bigram_terms = bigram_vectorizer.get_feature_names_out()

print("Original vocabulary size:", len(feature_names))
print("Optional-challenge vocabulary size:", len(bigram_terms))
print("Optional-challenge matrix shape:", bigram_matrix.shape)

bigram_df = pd.DataFrame(
    bigram_matrix.toarray(),
    index=titles,
    columns=bigram_terms
)

for title in bigram_df.index:
    top_terms = bigram_df.loc[title].sort_values(ascending=False).head(5)
    print(f"\n{title} — top terms/phrases")
    display(top_terms.rename("TF-IDF score").to_frame().round(4))
```

## 8. Final verification and short conclusion

**Verification:** This notebook checks that 10 documents are loaded, each document is longer than 30 words, the TF-IDF matrix has 10 rows, and all three expected query results rank first.

**Conclusion:** TF-IDF represents documents as weighted word features. Cosine similarity compares a query with those document vectors and ranks the closest matches. TF-IDF is useful for keyword-based search, but it does not fully understand meaning or word order.

```python
assert len(documents) == 10
assert all(len(document.split()) > 30 for document in documents)
assert tfidf_matrix.shape[0] == 10
assert all(rank_documents(q).iloc[0]["Document"] == expected
           for q, expected in zip(queries, expected_first))

print("SUCCESS: all required checks passed.")
```
